```python
#Simple KNN model Example:

from sklearn.neighbors import KNeighborsClassifier
import pandas as pd
d={"x1":[0.5,1,1.5,2],"x2":[0.5,1.2,2.5,2.5],"y":["good","good","bad","bad"]}
df=pd.DataFrame(d)
print(df)
x=df.iloc[:,0:2]
y=df.iloc[:,-1]
knn=KNeighborsClassifier(n_neighbors=3)
knn.fit(x,y)                               #Model Training
new_data_point = pd.DataFrame([[0.5, 0.5], [1.4, 1.9]], columns=x.columns)      # To avoid the 'feature names' warning, provide the input for prediction as a DataFrame with the same column names as 'x'
print(knn.predict(new_data_point))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.metrics import classification_report
import pandas as pd
df=pd.read_csv("/content/diabetes.csv")
print(df.shape)
print(df.isna().sum())

x=df.iloc[:,0:8].values    #.values are used to convert into array
y=df.iloc[:,-1].values

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=42)

mm=MinMaxScaler()
mm=MinMaxScaler()
mm.fit(x_train)                 #fit is used to find the xmax and xmin values in x_train.
x_train=mm.transform(x_train)   #Transform is used to apply minmax eq to all elements in the x_train and transform each elements by applying minmax eq to them.
x_test=mm.transform(x_test)     #Here we don't fit the test data cz to prevent data leaking ie, test data is used to check the results m/c provide.

knn=KNeighborsClassifier(n_neighbors=17)    #n_neighbors are called Parameters- Hyper Parameters Tuning-->Finding best values for parameters
knn.fit(x_train,y_train)     #Training/Learning

y_predict=knn.predict(x_test)

print("Accuracy of the Model: ",accuracy_score(y_test,y_predict)*100,"%")

print(ConfusionMatrixDisplay.from_predictions(y_test,y_predict))

print(classification_report(y_test,y_predict))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.metrics import classification_report
import pandas as pd
df=pd.read_csv("/content/lung_cancer_examples.csv")
#print(df.shape)
#print(df.isna().sum())

x=df.iloc[:,2:5].values    #.values are used to convert into array
y=df.iloc[:,-1].values

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=42)

mm=MinMaxScaler()
mm=MinMaxScaler()
mm.fit(x_train)                 #fit is used to find the xmax and xmin values in x_train.
x_train=mm.transform(x_train)   #Transform is used to apply minmax eq to all elements in the x_train and transform each elements by applying minmax eq to them.
x_test=mm.transform(x_test)     #Here we don't fit the test data cz to prevent data leaking ie, test data is used to check the results m/c provide.

knn=KNeighborsClassifier(n_neighbors=5)    #n_neighbors are called Parameters- Hyper Parameters Tuning-->Finding best values for parameters
knn.fit(x_train,y_train)     #Training/Learning

y_predict=knn.predict(x_test)

print("Accuracy of the Model: ",accuracy_score(y_test,y_predict)*100,"%")

print(ConfusionMatrixDisplay.from_predictions(y_test,y_predict))

print(classification_report(y_test,y_predict))
```
